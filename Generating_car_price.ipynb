```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

car_data = [
    ['BMW M3', 73200, 7.5],
    ['BMW 5 Series', 50200, 9.8],
    ['BMW X5', 12500, 13.2],
    ['BMW i8', 4700, 15.1],
    ['BMW X3', 61000, 8.2],
    ['BMW 7 Series', 39000, 10.3],
    ['BMW X6', 20500, 11.7],
    ['BMW 3 Series', 78200, 6.3],
    ['BMW Z4', 54500, 9.0],
    ['BMW M5', 22300, 11.9]

]

df = pd.DataFrame(car_data, columns = ['Car Name', "Speed", "Amount"])

# Select "Speed" as the independent variable (features) and reshape for scikit-learn
X = df[["Speed"]]
# Select "Amount" as the dependent variable (target)
y = df["Amount"]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size= 0.8, random_state= 40) # Increased train size for better model training
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))


# Initialize and train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

print("Actual test values:", list(y_test))
print("Predicted test values:", list(y_pred))

# Evaluate the model
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse:.2f}")
# Accessing coefficients and intercept
print(f"Slope (coef): {model.coef_[0]:.4f}")
print(f"Intercept: {model.intercept_:.2f}")


# New cars with different mileage for prediction
new_speed_data = [
    [10000],
    [50000]
]

# Make predictions on the new data
new_predictions = model.predict(new_speed_data)

for speed, price in zip([10000, 50000], new_predictions):
    print(f"Car with {speed} km mileage is predicted to cost ₦{price:.2f}M")


# Scatter plot of original data
plt.scatter(df['Speed'], df['Amount'], color='blue', label="Actual Data")

# Regression line
# Generate a range of speed values for plotting the regression line
speed_range = np.linspace(df['Speed'].min(), df['Speed'].max(), 100).reshape(-1, 1)
# Predict the corresponding 'Amount' values using the trained model
amount_pred = model.predict(speed_range)
plt.plot(speed_range, amount_pred, color='red', linewidth=2, label="Regression Line")

# Labels
plt.xlabel("Mileage (km)")
plt.ylabel("Price (₦ millions)")
plt.title("Car Price Prediction with Linear Regression")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
